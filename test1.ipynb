```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

# 1. Création d'un mini-dataset (Simulons des logs d'un portail captif)

```python
data = {
    'ip_address': ['192.168.1.10', '192.168.1.11', '192.168.1.12', '192.168.1.10', '192.168.1.13'],
    'octets_emis': [1500, 2300, 800, 4500, 1200],
    'duree_sec': [120, 450, 30, 600, 150],
    'type_appareil': ['Mobile', 'Laptop', 'Mobile', 'Laptop', 'IoT']
}
```

# Transformation en DataFrame (L'objet de base en Data Science)

```python
df = pd.DataFrame(data)
```

```python
print(df.info())
```

# 2. Affichage des premières lignes

```python
print("--- Aperçu du tableau ---")
print(df.head())
```

# 3. Statistiques descriptives (La magie commence ici)

```python
print("\n--- Analyse Statistique ---")
print(df.describe()) # Donne moyenne, min, max, écart-type automatiquement
```

# Groupement par type d'appareil et calcul de la moyenne de consommation

```python
consommation_moyenne = df.groupby('type_appareil')['octets_emis'].mean()
print("\n--- Consommation moyenne par appareil ---")
print(consommation_moyenne)
```

# Créer un graphique à barres pour la consommation par appareil

```python
df.plot(kind='bar', x='type_appareil', y='octets_emis', color='skyblue')

plt.title('Consommation par Type d\'Appareil')
plt.ylabel('Octets')
plt.show()
```
