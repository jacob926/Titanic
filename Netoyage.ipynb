```python
import pandas as pd
```

```python
data = {
    'date': ['2026-03-01', '2026-03-02', '2026-03-03', 'non-disponible', '2026-03-05'],
    'utilisateur': ['Etudiant_A', 'Etudiant_B', None, 'Etudiant_D', 'Etudiant_E'],
    'consommation_mo': [150, 2300, 45, 120, 9999999], # 9999999 est une aberration (outlier)
    'appareil': ['PC', 'PC', 'Smartphone', 'Smartphone', 'PC']
}

df = pd.DataFrame(data)
print("Données brutes et sales")
print(df)
```

```python
print(df.describe())
```

```python
# Supprimer les lignes où l'utilisateur est inconnu
df = df.dropna(subset=['utilisateur'])
print(df)
```

```python
print(df.info())
```

```python
# Convertir en date, les erreurs deviennent des "NaT" (Not a Time)
df['date'] = pd.to_datetime(df['date'], errors='coerce')

print(df.info())
```

```python
# On ne garde que les consommations réalistes (ex: < 10 000 Mo)
df = df[df['consommation_mo'] < 10000]

print(df.head())
```

```python
print("\n--- Données propres ---")
print(df)
print("\n--- Infos techniques ---")
print(df.info())
```

Type de donnée                          Problème,                           Solution recommandée

"Identifiant (ID, IP, Nom)",            Inconnu,                            Supprimer (On ne peut pas inventer une identité).

Série Temporelle (Date),                Trou dans la suite,                 ffill ou Interpolation.

"Mesure Physique (Temp, Poids)",        Capteur HS,                         Interpolation linéaire.

"Catégorie (Sexe, Ville, Appareil)",    Manquant,                           Mode (La plus fréquente).

Score / Note (0 à 20),                  Absent,                             Médiane (Pour éviter les notes extrêmes).
